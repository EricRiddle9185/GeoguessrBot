# ResNet-50 country classifier

Use the project Python environment. This notebook trains a country classifier on the frozen encoder and saves a new run.

Reference: [PyTorch transfer learning tutorial](https://docs.pytorch.org/tutorials/beginner/transfer_learning_tutorial.html).

```python
from pathlib import Path

from IPython.display import display
from PIL import Image

from train import build_classifier, make_parser, run_training
```

```python
number_of_countries = 10
hidden_sizes = [256]
dropout = 0.3
```

```python
classifier = build_classifier(2048, number_of_countries, hidden_sizes, dropout)
classifier
```

## Train the classifier

The preview above builds only the classifier. `run_training` loads and freezes ResNet-50, prepares the grouped data split, extracts features, and trains the classifier.

The default run samples up to 200 images from each of the ten largest country folders. Set `hidden_sizes` to `[]` for a linear classifier, `[256]` for one hidden layer, or `[256, 64]` for two. Each execution saves a new run under `../runs/training`.

```python
args = make_parser().parse_args([
    '--num-countries', str(number_of_countries),
    '--max-per-country', '200',
    '--hidden-sizes', *[str(size) for size in hidden_sizes],
    '--dropout', str(dropout),
    '--epochs', '30',
    '--seed', '416',
])
result = run_training(args)
result
```

```python
run_dir = Path(result['run_directory'])
display(Image.open(run_dir / 'training_curves.png'))
display(Image.open(run_dir / 'confusion_matrix.png'))
```
