# Country classifier

Train a classifier using frozen ResNet-50 features.

Reference: [PyTorch transfer learning tutorial](https://docs.pytorch.org/tutorials/beginner/transfer_learning_tutorial.html).

```python
from pathlib import Path

from IPython.display import display
from PIL import Image

from train import build_classifier, make_parser, run_training
```

```python
number_of_countries = 10
hidden_sizes = [256]
dropout = 0.3
```

```python
classifier = build_classifier(2048, number_of_countries, hidden_sizes, dropout)
classifier
```

## Training

Defaults: ten countries, up to 200 images per country, and 30 epochs.

Use `[]` for a linear classifier, `[256]` for one hidden layer, or `[256, 64]` for two. Runs are saved in `../runs/training`.

```python
args = make_parser().parse_args([
    '--num-countries', str(number_of_countries),
    '--max-per-country', '200',
    '--hidden-sizes', *[str(size) for size in hidden_sizes],
    '--dropout', str(dropout),
    '--epochs', '30',
    '--seed', '416',
])
result = run_training(args)
result
```

```python
run_dir = Path(result['run_directory'])
display(Image.open(run_dir / 'training_curves.png'))
display(Image.open(run_dir / 'confusion_matrix.png'))
```
